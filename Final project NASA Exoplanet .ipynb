**Type of Analysis**

**1. Descriptive Analysis**

Descriptive analysis summarizes the dataset using statistical measures such as mean, count, and distribution. It helps in understanding the overall structure and key characteristics of planetary and stellar properties. This provides a clear overview of the data before deeper analysis.

**2. Diagnostic Analysis**

Diagnostic analysis focuses on identifying relationships and the reasons behind observed patterns in the data. It uses techniques like correlation, grouping, and comparisons to answer “why” something occurred. For example, it explains why certain discovery methods are more successful than others.

**3. Predictive Analysis**

Predictive analysis uses historical data to identify trends and forecast future outcomes. It applies simple trend analysis or basic models to estimate how exoplanet discoveries may increase over time. This helps in anticipating future patterns based on past data.

**4. Prescriptive Analysis**

Prescriptive analysis provides recommendations based on insights derived from the data. It suggests possible actions or strategies, such as focusing on effective detection methods or favorable planetary conditions. This supports decision-making and improves future outcomes.

**Stage 2 - Data Cleaning and Pre-processing**

```python
numerical_columns = ['disc_year','pl_bmasse','pl_rade','pl_orbper',
                     'pl_orbsmax','pl_orbeccen','pl_eqt',
                     'st_mass','st_rad','st_teff',
                     'st_lum','sy_dist']

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].mean())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['disc_year'] = df['disc_year'].astype(int)
# disc_year was originally stored as float datatype.
# It is converted into integer because year values
# should not contain decimal points.
```

```python
df['planet_star_ratio'] = df['pl_orbsmax'] / df['st_rad']
# Creating new derived feature:
# planet_star_ratio compares orbital distance
# with host star radius.
```

```python
df['mass_radius_ratio'] = df['pl_bmasse'] / df['pl_rade']
# Creating another derived feature:
# mass_radius_ratio helps analyze planetary density
# and structural characteristics.
```

```python
df.head()
```

**Outlier Detection**

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df['pl_bmasse'])

plt.title("Outlier Detection for Planet Mass")

plt.show()# Boxplot is used to identify extreme values (outliers)
# in planet mass distribution.
```

**Outlier Treatment Code**

```python
Q1 = df['pl_bmasse'].quantile(0.25)
Q3 = df['pl_bmasse'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[                               # Keeping only values within lower and upper bounds
    (df['pl_bmasse'] >= lower_bound) &
    (df['pl_bmasse'] <= upper_bound) # IQR (Interquartile Range) method is used
# to detect and remove extreme outliers.
]
```

**Check Skewness**

```python
print(df['pl_bmasse'].skew()) # Skewness measures whether data distribution
                              # is balanced or highly uneven.
```

**Apply Log Transformation**

```python
df['log_pl_bmasse'] = np.log1p(df['pl_bmasse']) #Log transformation is applied to reduce skewness and normalize highly uneven distributions.

                                                #It improves data balance and analysis quality.
```

**BARPLOT**

```python
avg_radius = df.groupby('discoverymethod')['pl_rade'].mean().sort_values(ascending=False)

plt.figure(figsize=(12,6))

sns.barplot(
    x=avg_radius.index,
    y=avg_radius.values
)

plt.title("Average Planet Radius by Discovery Method")
plt.xlabel("Discovery Method")
plt.ylabel("Average Planet Radius")

plt.xticks(rotation=90)

plt.show()
```

**Interpretation of Above Chart**

planet detection technologies used in astronomy.
