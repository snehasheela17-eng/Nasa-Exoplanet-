## 🧹 Stage 2 – Data Cleaning and Pre-processing


```
# ── Step 1: Impute Missing Values with Column Mean ────────────────────────────
rows_before = len(df)

numerical_columns = ['disc_year', 'pl_bmasse', 'pl_rade', 'pl_orbper',
                     'pl_orbsmax', 'pl_orbeccen', 'pl_eqt',
                     'st_mass', 'st_rad', 'st_teff', 'st_lum', 'sy_dist']

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].mean())

print(f"✅ Missing values imputed using column mean")
print(f"   Total missing values now: {df.isnull().sum().sum()}")
```

```
# ── Step 2: Confirm No Duplicates ─────────────────────────────────────────────
print(f"Duplicate rows: {df.duplicated().sum()}")
```

```
# ── Step 3: Fix Data Types ────────────────────────────────────────────────────
# disc_year was float due to NaN — convert to int after imputation
df['disc_year'] = df['disc_year'].astype(int)
print(f"✅ disc_year converted to integer (was float64 due to NaN values)")
print(df['disc_year'].dtype)
```

```
# ── Step 4: Feature Engineering (Derived Columns) ────────────────────────────
# Ratio of orbital distance to star radius — shows how close a planet orbits its star
df['planet_star_ratio'] = df['pl_orbsmax'] / df['st_rad']

# Mass-to-radius ratio — proxy for planetary density
df['mass_radius_ratio'] = df['pl_bmasse'] / df['pl_rade']

print("✅ New derived features created:")
print(f"   planet_star_ratio  → orbital distance / star radius")
print(f"   mass_radius_ratio  → planet mass / planet radius (density proxy)")
df[['pl_name', 'planet_star_ratio', 'mass_radius_ratio']].head()
```

```
# ── Step 5: Outlier Detection (Boxplot) ───────────────────────────────────────
fig, ax = plt.subplots(figsize=(9, 3))
sns.boxplot(x=df['pl_bmasse'], color='#7F77DD', ax=ax)
ax.set_title('Outlier Detection — Planet Mass (pl_bmasse)', fontsize=13, pad=10)
ax.set_xlabel('Planet Mass (Earth Masses)', fontsize=10)
plt.tight_layout()
plt.show()
```

**Interpretation:** The boxplot shows extreme outliers on the right side (very massive planets). These outliers will be removed using the IQR method to avoid skewing the analysis.

```
# ── Step 6: Outlier Treatment using IQR ──────────────────────────────────────
rows_before_outlier = len(df)

Q1 = df['pl_bmasse'].quantile(0.25)
Q3 = df['pl_bmasse'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df['pl_bmasse'] >= lower_bound) & (df['pl_bmasse'] <= upper_bound)]

print(f"✅ Outlier Treatment Complete (IQR Method)")
print(f"   Rows before : {rows_before_outlier:,}")
print(f"   Rows after  : {len(df):,}")
print(f"   Removed     : {rows_before_outlier - len(df):,} outlier rows")
print(f"   Lower bound : {lower_bound:.2f}")
print(f"   Upper bound : {upper_bound:.2f}")
```

```
# ── Step 7: Skewness Check ────────────────────────────────────────────────────
skew_cols = ['pl_bmasse', 'pl_rade', 'pl_orbper', 'sy_dist']
skew_vals = df[skew_cols].skew().round(3)
print("Skewness values (>1 or <-1 = significantly skewed):")
for col, val in skew_vals.items():
    flag = "⚠️ High skew" if abs(val) > 1 else "✅ Acceptable"
    print(f"   {col:<15} : {val:>8}  {flag}")
```

```
# ── Step 8: Log Transformation to Reduce Skewness ────────────────────────────
df['log_pl_bmasse'] = np.log1p(df['pl_bmasse'])

print(f"Skewness BEFORE log transform : {df['pl_bmasse'].skew():.4f}")
print(f"Skewness AFTER  log transform : {df['log_pl_bmasse'].skew():.4f}")
print("✅ Log transformation applied — skewness significantly reduced")
```

```
# ── Step 9: Save Cleaned Dataset ─────────────────────────────────────────────
df.to_csv('cleaned_exoplanet_data.csv', index=False)
print(f"✅ Cleaned dataset saved as 'cleaned_exoplanet_data.csv'")
print(f"   Final shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
df.head()
```

**Interpretation — Planet Mass Distribution (Histogram + KDE)**

- **Feature used:** `pl_bmasse` (planet mass in Earth masses)
- **What it shows:** The distribution of planet masses across all confirmed exoplanets in the dataset
- **Key insight:** The distribution is right-skewed — the vast majority of detected exoplanets have relatively low masses (clustered near 0–5 Earth masses), while very massive planets are rare. This reflects both a physical reality (small planets are more common) and an observational bias (massive planets are easier to detect, so the far tail is still populated). The KDE curve confirms the sharp peak at low masses.

The main cluster confirms that most discovered planets are "super-Earths" or "mini-Neptunes" in size, which aligns with what Kepler and TESS missions have consistently found.

**Interpretation — Planet Count by Discovery Method (Countplot)**

- **Chart type:** Horizontal bar chart (countplot)
- **Feature used:** `discoverymethod`
- **What it shows:** How many exoplanets have been discovered using each detection technique
- **Key insight:** The **Transit method** dominates overwhelmingly — it accounts for over 3,800 discoveries, driven by Kepler and TESS space telescopes which monitor thousands of stars simultaneously. **Radial Velocity** is a distant second (~1,000). This reflects both the effectiveness and scalability of the transit technique. Rarer methods like Imaging and Timing Variations are technologically limited in scope.


### Bivariate Analysis


```
# ── Chart 4: Scatterplot — Planet Mass vs Radius ──────────────────────────────
fig, ax = plt.subplots(figsize=(9, 6))
sc = ax.scatter(df['pl_bmasse'], df['pl_rade'],
                alpha=0.45, s=18, color='#7F77DD', edgecolors='none')

# Add trend line
z = np.polyfit(df['pl_bmasse'], df['pl_rade'], 1)
p = np.poly1d(z)
x_line = np.linspace(df['pl_bmasse'].min(), df['pl_bmasse'].max(), 200)
ax.plot(x_line, p(x_line), '--', color='#EF9F27', linewidth=2, label='Trend line')

ax.set_title('Planet Mass vs Planet Radius', fontsize=14, pad=14, color='white')
ax.set_xlabel('Planet Mass (Earth Masses)', fontsize=11)
ax.set_ylabel('Planet Radius (Earth Radii)', fontsize=11)
ax.legend(fontsize=9, facecolor='#1a1a2e', edgecolor='#444', labelcolor='white')
plt.tight_layout()
plt.savefig('chart4_mass_radius_scatter.png', dpi=150, bbox_inches='tight', facecolor='#0E1117')
plt.show()
```

**Interpretation — Planet Mass vs Planet Radius (Scatterplot)**

- **Chart type:** Scatterplot with trend line
- **Features used:** `pl_bmasse` (mass), `pl_rade` (radius)
- **What it shows:** The relationship between a planet's mass and its size
- **Key insight:** There is a **positive correlation** — heavier planets tend to be larger. However, the scatter is significant, meaning many planets deviate from this trend. Some high-mass planets have small radii (dense, rocky composition) while some low-mass planets have large radii (puffy, gaseous interiors). This diversity reveals that **composition**, not just mass, drives planetary size.

Transit and Radial Velocity methods, while most prolific in count, detect a broader range including smaller planets.
```
# ── Chart 6: Correlation Heatmap ─────────────────────────────────────────────
cols_for_corr = ['pl_bmasse', 'pl_rade', 'pl_orbper', 'pl_orbsmax',
                 'pl_orbeccen', 'st_mass', 'st_teff', 'st_lum', 'sy_dist']
labels = ['Mass', 'Radius', 'Orb.Period', 'Orb.Dist',
          'Eccentricity', 'StarMass', 'StarTemp', 'StarLum', 'Distance']

corr_df = df[cols_for_corr].copy()
corr_df.columns = labels
corr = corr_df.corr()

import numpy as np
mask = np.triu(np.ones_like(corr, dtype=bool))

fig, ax = plt.subplots(figsize=(11, 8))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, vmin=-1, vmax=1, ax=ax,
            linewidths=0.4, linecolor='#111111',
            annot_kws={'size': 9, 'color': 'white'},
            cbar_kws={'shrink': 0.8})
ax.set_title('Feature Correlation Heatmap', fontsize=14, pad=14, color='white')
ax.tick_params(labelsize=9, colors='#cccccc')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('chart6_heatmap.png', dpi=150, bbox_inches='tight', facecolor='#0E1117')
plt.show()
```

**Interpretation — Feature Correlation Heatmap**

### 📋 Summary of Findings — What the Patterns Mean

**1. Transit Method is the King of Discovery**
The Transit method accounts for over 75% of all confirmed exoplanet discoveries. This is entirely because of space telescopes like Kepler (2009) and TESS (2018) which observe hundreds of thousands of stars simultaneously, detecting tiny dips in starlight as planets cross in front of them.

**2. Small Planets Are Far More Common**
Over 90% of exoplanets in the dataset have masses below 10 Earth masses. This suggests the universe overwhelmingly produces small, rocky or gaseous worlds rather than massive gas giants. However, this also reflects observational bias — massive planets are easier to spot.

**3. Technology Drives Discovery**
Discovery counts show sharp spikes in 2014 and 2016 (Kepler bulk confirmations) and again after 2019 (TESS data releases). This shows that astronomical discovery is fundamentally mission-driven — budget and technology, not natural discovery rate, determine how many planets we find.

**4. Hot Jupiters Are Physically Extreme**
The bubble chart reveals a cluster of high-mass, large-radius, high-temperature planets with short orbital periods — these are "Hot Jupiters." Their physical properties are interconnected: massive planets close to their stars get baked by radiation, inflating their atmospheric radii.

**5. Stellar and Planetary Properties Are Linked**
The correlation heatmap confirms that star luminosity and temperature are strongly correlated (as stellar physics predicts), and that orbital period and orbital distance follow Kepler's Third Law. This validates data integrity and shows the physical universe follows predictable rules even in datasets with thousands of worlds.

### 💡 Recommendations for Decision-Making

1. **Invest in transit-based space telescopes** — the ROI in discovery volume is unmatched vs ground-based methods. Future missions should continue this approach.
2. **Prioritize follow-up studies on mid-mass planets** (5–15 Earth masses) — these "Super-Earths" and "Mini-Neptunes" are abundant but poorly understood compositionally.
3. **Develop better ground-based imaging techniques** — Imaging currently finds the fewest planets but detects the most unusual, massive worlds unavailable to other methods.
4. **Use mass-radius ratio as a planetary density classifier** — the derived `mass_radius_ratio` feature can help categorize rocky vs gaseous planets without spectroscopy.
5. **Expand observations in the 2–10 AU orbital range** — current detection bias skews heavily toward close-in planets. Planets in habitable zones (0.5–2 AU) remain underrepresented.
